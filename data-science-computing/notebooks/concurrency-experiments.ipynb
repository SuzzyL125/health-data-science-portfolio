```python
#Homework2-ZLi
#q1:
import threading
from threading import Barrier
import random
import time

class ProducerConsumer:
    def __init__(self):
        self.items = []
        self.lock = threading.Lock()
        self.barrierwait = threading.Barrier(4)  


    def producer(self):
        self.lock.acquire()
        for _ in range(5):
            self.items.append(random.randint(1,100))
        print(f"Current list: {self.items}")
        self.lock.release()
        self.barrierwait.wait() 

    def consumer(self):
        self.barrierwait.wait() 
        self.lock.acquire()
        for _ in range(5):
            if self.items:
                random_index = random.randint(0, len(self.items) - 1)
                self.items.pop(random_index)
        print(f"Current list: {self.items}")
        self.lock.release()


def main():
    pc = ProducerConsumer()
    print('Program starts')
    producer_thread1 = threading.Thread(target=pc.producer)
    producer_thread2 = threading.Thread(target=pc.producer)
    consumer_thread1 = threading.Thread(target=pc.consumer)
    consumer_thread2 = threading.Thread(target=pc.consumer)

    producer_thread1.start()
    producer_thread2.start()
    consumer_thread1.start()
    consumer_thread2.start()

    producer_thread1.join()
    producer_thread2.join()
    consumer_thread1.join()
    consumer_thread2.join()

    print(print("\nProgram ends"))

if __name__ == "__main__":
    main()
```

```python
import random
import time
import threading
import numpy
import copy
# Generate random arrays, pay attention that once arrary is sorted, the next function will use sorted array but we want to use the original one
array10k = numpy.array([random.randint(0, 10000) for _ in range(10000)])
array_10m = numpy.array([random.randint(0, 10000) for _ in range(10000000)])
array10k2=copy.deepcopy(array10k)
array10m2=copy.deepcopy(array_10m)
# Straight Insertion Sort
def insertion_sort(arr):
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key

# Function to sort array and save to binary file
# Sorting array of 10,000 integers
print('Non-threaded function starts.\n')
start_time_sorting = time.perf_counter()
insertion_sort(array10k)
sorting_time = time.perf_counter() - start_time_sorting

# Saving array of 10,000,000 integers to binary file
start_time_saving = time.perf_counter()
for i in range (100):
    file = open("c:/Users/zli21/Documents/Python/non_threaded2.bin", 'a+b')
    file.write(array_10m)
    file.close
saving_time = time.perf_counter() - start_time_saving

total_time = time.perf_counter() - start_time_sorting
print("Sorting time:", sorting_time)
print("File saving time:", saving_time)
print('Non-threaded function ends.\n')
print("Total running time:", total_time)

# Function for threaded sorting and saving
# Sorting array of 10,000 integers
def sorting_thread(array):
    start_time_sorting = time.perf_counter()
    insertion_sort(array)
    sorting_time = time.perf_counter() - start_time_sorting
    print("Sorting time (Threaded):", sorting_time,"\n")

# Saving array of 10,000,000 integers to binary file
def saving_thread(array):
    start_time_saving = time.perf_counter()
    for i in range (100):
        file = open("c:/Users/zli21/Documents/Python/threaded2.bin", 'a+b')
        file.write(array)
        file.close
    saving_time = time.perf_counter() - start_time_saving
    print("Saving time (Threaded):", saving_time)

print("Threaded function starts.\n")
# Creating threads for sorting and saving
sorting_thread = threading.Thread(target=sorting_thread,args=[array10k2])
saving_thread = threading.Thread(target=saving_thread,args=[array10m2])

# Starting threads
start_time_total = time.perf_counter()
sorting_thread.start()
saving_thread.start()

# Waiting for threads to finish
sorting_thread.join()
saving_thread.join()

total_time = time.perf_counter() - start_time_total
print("Threaded function ends.\n")
print("Total running time (Threaded):", total_time)
```
